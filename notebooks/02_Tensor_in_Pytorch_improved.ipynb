# Введение в тензоры PyTorch

Этот ноутбук посвящен основам работы с тензорами в библиотеке PyTorch.

Тензор — это многомерный массив, аналогичный `numpy.ndarray`, но с <br>ключевыми преимуществами:
1. Возможность выполнения вычислений на GPU (видеокарте).
2. Встроенная поддержка автоматического дифференцирования (autograd) для <br>обучения нейросетей.

```python
import torch
import numpy as np
```

## 1. Создание тензоров

Основной способ создания тензора — функция `torch.tensor()`. Она является <br>аналогом `numpy.array()`.

```python
# Создание тензора из списка
t1 = torch.tensor([2, 3])
print(f"1D тензор {t1.dtype}:\n", t1, "\n")

# Создание 2D тензора с явным указанием типа данных
t2 = torch.tensor([[2, 3], [4, 5]], dtype=torch.float32)
print(f"2D тензор {t2.dtype}:\n", t2, "\n")

# Создание тензора с требованием вычисления градиента
t3 = torch.tensor([[2.0, 3.0], [4.0, 5.0]], requires_grad=True)
print(f"2D nензор {t3.dtype}:\n", t3)
```

<!-- [ИЗМЕНЕНО]: Добавлено пояснение к ошибке, чтобы это было обучающим моментом, а не просто крашем -->
### ⚠️ Важное правило для `requires_grad`
Градиенты можно вычислять **только** для тензоров с плавающей точкой (`float`) <br>или комплексных чисел. Целые числа (`int`) не поддерживают вычисление <br>градиентов.

```python
# Попытка создать целочисленный тензор с requires_grad=True вызовет ошибку:
t4 = torch.tensor([[2, 3], [4, 5]], requires_grad=True)
```

### Связь с NumPy
Мост между обработкой данных (Pandas/NumPy) и моделью (PyTorch).

```python
# Создание тензора из NumPy массива
np_arr = np.array([[1, 2], [3, 4]])
t_np = torch.from_numpy(np_arr)
print("Тензор PyTorch:\n", t_np)
```

```python
# Обратное преобразование
new_arr = t_np.numpy()
print("\nМассив Numpy:\n", new_arr)
```

```python
# ВАЖНО: torch.from_numpy() не копирует данные, они делят одну память!
t_np[0, 0] = 999
print("\nNumPy массив после изменения тензора:\n", np_arr) # Изменился и он!
```

```python
t_req_grad = torch.tensor([[2.0, 3.0], [4.0, 5.0]], requires_grad=True)

new_arr = t_req_grad.detach().numpy()
print("\nМассив Numpy:\n", new_arr)
```

## 2. Свойства и атрибуты тензоров

Важно уметь проверять размерность, тип данных и извлекать скалярные значения.

```python
tensor = torch.tensor(
    [[[2, 3], [4, 5]], [[6, 7], [8, 9]]],
    dtype=torch.float32,
    requires_grad=True
)

print("Тип данных (dtype):", tensor.dtype)
print("Форма (shape):", tensor.shape)       # Атрибут shape
print("Размер (size()):", tensor.size())     # Метод size()
print("Количество измерений (ndim):", tensor.ndim)
```

```python
# Пример базового среза (slicing), который используется постоянно
print("\nПервый 'слой' матрицы 2x2:\n", tensor[0])
print("Все элементы второго столбца первого слоя:\n", tensor[0, :, 1])
```

```python
# Индексация возвращает тензор, даже если это один элемент
element = tensor[0, 0, 0]
print("\nЭлемент с индексом [0, 0, 0]:", element)
print("Тип элемента:", type(element))

# Метод .item() извлекает стандартное число Python из тензора,
# содержащего ОДИН элемент
print("\nЗначение как число Python (.item()):", element.item())
print("Тип после .item():", type(element.item()))
```

## 3. Специальные функции создания тензоров

PyTorch предоставляет удобные функции для инициализации тензоров определенными <br>значениями (часто используется для создания весов или масок).

```python
# Тензор из нулей и единиц
zeros_tensor = torch.zeros([2, 3], dtype=torch.int32)
ones_tensor = torch.ones([2, 3])

# Создание тензора с формой и типом данных как у другого тензора (очень полезно!)
zeros_like_tensor = torch.zeros_like(ones_tensor)
full_like_tensor = torch.full_like(zeros_like_tensor, 7) # Заполняет семерками

print("full_like тензор:\n", full_like_tensor)

# Арифметическая прогрессия
print("\narange от 2 до 10 с шагом 0.5:\n", torch.arange(2, 10, 0.5))

# Диагональные и треугольные матрицы
print("\nДиагональная матрица:\n", torch.diag(torch.tensor([5, 4])))
print("\nЕдиничная матрица 5x5:\n", torch.eye(5))
```

## 4. Изменение формы (Reshape) и размерности

Часто требуется изменить форму тензора без изменения его данных (например, <br>"сплющить" изображение перед линейным слоем).

```python
x = torch.arange(12).view(3, 4)
print("Исходный тензор непрерывен:", x.is_contiguous())
print(x)
```

```python
# 1. view() - быстрый, но требовательный
y = x.view(4, 3)       # Работает отлично
z = x.view(-1)         # -1 значит "вычисли размерность сам" (здесь будет 12)

print("Размер тензора Y:", y.shape)
print("Размер тензора Z:", z.shape)
```

```python
# 2. Ловушка: transpose нарушает непрерывность памяти
x_t = x.T                # Транспонирование
print("Тензор x_t в памяти хранится непрерывно:", x_t.is_contiguous())  # False!

# ❌ Это упадет с ошибкой:
try:
    bad = x_t.view(12)
except RuntimeError as e:
    print("\nОшибка view:", str(e).split('\n')[0])

# ✅ Решение 1: сначала сделать непрерывным, потом view
new_x = x_t.contiguous().view(12)
print("\nТензор new_x непрерывен:", new_x.is_contiguous())
```

```python
# ✅ Решение 2 (РЕКОМЕНДУЕМОЕ): использовать reshape()
# reshape() умный: если память непрерывна -> работает как view (быстро, без копирования)
# если нет -> делает копию данных в новом порядке (медленнее, но гарантированно работает)
print("Тензор x_t в памяти хранится непрерывно:", x_t.is_contiguous())

new_x = x_t.reshape(12)
print("\nТензор new_x непрерывен:", new_x.is_contiguous())
print("Размер тензора new_x:", new_x.shape)
```

```python
new_x[5] = 9999

print("\nТензор x_t:\n", x_t) # Изменился и он!
print("\nТензор new_x:\n", new_x) # Изменился и он!
```

```python
# ✅ Решение 2 (РЕКОМЕНДУЕМОЕ): использовать reshape()
# reshape() умный: если память непрерывна -> работает как view (быстро, без копирования)
# если нет -> делает копию данных в новом порядке (медленнее, но гарантированно работает)
print("Тензор new_x в памяти хранится непрерывно:", new_x.is_contiguous())

Y = new_x.reshape(2, 6)
print("\nТензор Y непрерывен:", Y.is_contiguous())
print("Размер тензора Y:", Y.shape)

Y[1, 3] = 7777

print("\nТензор new_x:\n", new_x) # Изменился и он!
print("\nТензор Y:\n", Y) # Изменился и он!
```

## 5. Базовые математические операции <br>и Broadcasting

<!-- [ДОБАВЛЕНО]: Раздел про Broadcasting, так как это фундаментальная концепция, отсутствовавшая в оригинале -->

```python
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5., 6.], [7., 8.]])

# Поэлементные операции
print("Сложение:\n", a + b)
print("Поэлементное умножение (НЕ матричное!):\n", a * b)
print("Возведение в квадрат:\n", a ** 2)

# Матричное умножение (основа нейросетей)
print("\nМатричное умножение (оператор @):\n", a @ b)
print("Матричное умножение (функция):\n", torch.matmul(a, b))

# Агрегация
print("\nСреднее значение всех элементов:", a.mean().item())
print(
    "Сумма по столбцам с сохранением размерности (dim=0, keepdims=True):\n",
    a.sum(dim=0, keepdims=True)
)
print("", a.sum(dim=0, keepdims=True).shape)
```

### 📡 Broadcasting (Транслирование)
Если размерности не совпадают, PyTorch пытается "растянуть" меньший тензор до <br>размера большего без реального копирования данных в памяти.

```python
matrix = torch.tensor([[1, 2], [3, 4], [5, 6]]) # Форма: 3x2
vector = torch.tensor([10, 20])                 # Форма: 1D, длина 2

# PyTorch автоматически "размножит" vector на каждую строку matrix
print("Результат Broadcasting (matrix + vector):\n", matrix + vector)
# Эквивалентно: [[1+10, 2+20], [3+10, 4+20], [5+10, 6+20]]
```

## 6. Индексация и булевы маски

В машинном обучении часто нужно отфильтровать данные по условию (например, <br>найти все потери (loss), которые больше определенного порога).

```python
scores = torch.tensor([85, 42, 90, 33, 78])

# 1. Создаем булеву маску (True там, где условие выполняется)
mask = scores > 50
print("Маска (scores > 50):\n", mask)

# 2. Применяем маску для фильтрации
high_scores = scores[mask]
print("\nВысокие оценки:\n", high_scores)

# 3. Fancy indexing (выбор конкретных индексов)
indices = torch.tensor([0, 2, 4])
print("\nЭлементы по индексам 0, 2, 4:\n", scores[indices])
```

## 7. Граф вычислений (Autograd)

PyTorch автоматически строит граф вычислений для тензоров с `requires_grad=True`. <br>Это основа алгоритма обратного распространения ошибки (backpropagation).

> ⚠️ **Примечание:** Для визуализации графа требуется библиотека `torchviz`.

```python
# Раскомментируйте строку ниже, если torchviz не установлен
# !pip install torchviz
```

```python
from torchviz import make_dot

# Пример 1: Полный граф вычислений
x = torch.ones(10, requires_grad=True)
y = 2 * x
z = 3 + x
t = 5 * z
w = t + y
r = (y + w).sum()

print("Визуализация полного графа вычислений для r:")
make_dot(r)
```

```python
# Пример 2: Использование .detach()
# Метод .detach() "открепляет" тензор от графа вычислений.
# По нему не будет распространяться градиент, что экономит память и ускоряет вычисления,
# если этот промежуточный результат не нужен для обучения.

x = torch.ones(10, requires_grad=True)
y = 2 * x
z = 3 + x.detach()  # <--- x откреплен, градиент через z на x не пойдет
t = 5 * z
w = t + y
r = (y + w).sum()

print("Визуализация графа с использованием .detach():")
print("Обратите внимание, что ветка, идущая от x к z, исчезла из графа градиентов.")
make_dot(r)
```

## 8. Копирование тензоров

При работе с градиентами важно понимать, копируются ли данные и сохраняется ли <br>история вычислений.

```python
x = torch.tensor([1.0, 2.0], requires_grad=True)
y = x * 2  # y требует градиент, так как x требует градиент

print("Исходный Y:", y)
print("Y требует градиент:", y.requires_grad)
```

```python
# 1. clone() - копирует данные и историю градиентов
y_clone = y.clone()
print("y.clone() требует градиент:", y_clone.requires_grad) # True
```

```python
# 2. detach() - открепляет от графа (requires_grad=False), но ДЕЛИТ память с оригиналом!
# Изменение y_detached изменит и y, что может сломать вычисление градиентов.
y_detached = y.detach()
print("Y требует градиент:", y.requires_grad) # True
print("y.detach() требует градиент:", y_detached.requires_grad) # False
```

```python
y_detached[0] = 9999.
print("\nТензор y_detached:\n", y_detached) # Изменился и он!
print("\nТензор Y:\n", y) # Изменился и он!
```

```python
# 3. clone().detach() или detach().clone() - БЕЗОПАСНАЯ КОПИЯ.
# Создает новый тензор в новой области памяти БЕЗ истории градиентов.
# Это рекомендуемый способ, если вам нужно "заморозить" тензор для логирования или метрик.
y_safe = y.clone().detach()
print("y.clone().detach() требует градиент:", y_safe.requires_grad) # False
```

```python
# Доказательство разделения памяти:
y_safe[1] = 777.0
print("\nПосле изменения y_safe, исходный y остался прежним:\n", y)
```

## 9. Работа с устройствами: CPU и GPU

Одна из главных причин использования PyTorch — возможность легко переносить <br>вычисления на видеокарту (GPU) для ускорения.

```python
# Проверка доступности устройств
print("Доступен ли CPU:", torch.cpu.is_available()) # Всегда True
print("Доступен ли GPU (CUDA):", torch.cuda.is_available())
```

```python
# ❌ Плохой стиль (использование строк, сложно масштабировать):
# device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ✅ Хороший стиль (объект torch.device, рекомендуемый PyTorch):
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print("\nИспользуемое устройство:", device)
```

```python
if device.type == 'cuda':
    idx = torch.cuda.current_device()
    print(f"device.index: {idx}")
    props = torch.cuda.get_device_properties(idx)
    print(f"Название: {props.name}")
    print(f"Всего памяти: {props.total_memory / 1024**3:.1f} ГБ")

    # Свободная и общая память (кортеж: free, total)
    free_mem, total_mem = torch.cuda.mem_get_info(idx)
    print(f"Свободно памяти: {free_mem / 1024**3:.1f} ГБ")
    print(f"Занято памяти: {(total_mem - free_mem) / 1024**3:.1f} ГБ")

    # Версии
    print(f"Версия CUDA: {torch.version.cuda}")
```

```python
# Создание тензора и его перенос на выбранное устройство

# Тензор сразу на целевом устройстве
tensor_gpu = torch.tensor([1., 2., 3.], requires_grad=True, device=device)
print("Тензор 'tensor_gpu' на устройстве:", tensor_gpu.device)

# Перенос существующего тензора
tensor_cpu = torch.tensor([1., 2., 3.])
tensor_moved = tensor_cpu.to(device)
print("Тензор 'tensor_moved' на устройстве:", tensor_moved.device)
```

⚠️ **Золотое правило:** Операции можно выполнять **только** между тензорами, <br>находящимися на ОДНОМ устройстве!

```python
# ❌ НЕправильно: вызовет ошибку RuntimeError
result = tensor_gpu + tensor_cpu
```

```python
# ✅ Правильно: привести оба к одному устройству
result = tensor_gpu + tensor_cpu.to(device)
print("\nРезультат сложения:", result)
print("Устройство результата:", result.device)
```

```python
# 🏆 ЛУЧШАЯ ПРАКТИКА извлечения данных из тензора с градиентом на GPU

# 1. .detach() - убираем из графа вычислений (иначе numpy() выдаст ошибку)
# 2. .cpu() - переносим данные из видеопамяти в оперативную память (обязательно перед numpy!)
# 3. .numpy() - конвертируем в массив NumPy

numpy_ready_tensor = result.detach().cpu().numpy()
print("\nТип после detach().cpu().numpy():", type(numpy_ready_tensor))
print("Значение:", numpy_ready_tensor)
```
